```python
#String is heterogeneous. Save multiple data at one time. 
mark = ["karan","om",90,56,55] 
print(len(mark))
mark[0] = "om"
print(mark)
```

```python
#Slicing in lists.
list = [95, 94, 87, 64, 65] 
print(list[2:4])
print(list[-4:-1])
```

```python
#List method or functions: append, sort, insert
list = [95, 94, 87, 64, 65] 
list.append(33) # Append use or add the element in a lastly. 
print(list)
```

```python
## Sort method 
list = [95, 94, 87, 64, 65] 
list.sort()#Sort method used for ascending the number. 
print(list)
```

```python
## reverse  sort method
list = [95, 94, 87, 64, 65] 
list.sort(reverse = True) #Reverse method used for descending order. 
print(list)
```

```python
list = ["Apple", "banana", "pineapple", "grapes"] 
list.sort()#These methods are used for accessing alphabetical order. 
print(list)
```

```python
# Reverse  matter.
list = [95, 94, 87, 64, 65] 
list.reverse()#reverse method used for reverseing the number. 
print(list)

```

```python
#Insert method
list = [95, 94, 87, 64, 65] 
list.insert(2,67) # Insert method used for adding numbers in through the index
print(list)
```

```python
# remove method
list = [95, 94, 87, 64, 65] 
list.remove(64) # remove method used for delete the specific no.
print(list)

```

```python
# pop method
list = [95, 94, 87, 64, 65] 
list.pop(4) #  pop method used for deleting the value using index
print(list)
```
